## Using Transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer(transformers=[
    ("tnf1",OrdinalEncoder(categories=[["Mild","Strong"]]),["cough"]),
    ("tnf2",SimpleImputer(),["fever"]),
    ("tnf3",OneHotEncoder(drop="first",sparse_output=False,dtype=int),["gender","city"])
],remainder="passthrough")
```

### SO basically ColumnTransformer does that every thing written above like for ordinal encoding and all it took very long time as we had to do each and every column seprate on the basis operation needed to perform and then we needed to merge the whole transformed data later is written in one very line. This used very less lines of code and saved time.

### In ColumnTransformer we use transformer where we mention which all operation we need to do, and in Reminder we either write "drop" or "passthrough" which means that thing decides weather the column which was not used for transforming or operation should be droped or passed through in the example we had age which we passthrough.
