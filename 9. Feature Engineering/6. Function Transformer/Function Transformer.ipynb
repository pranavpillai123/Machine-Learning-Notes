## How to plot QQplt

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("iris")
```

```python
df.head()
```

```python
temp=sorted(df['sepal_length'].tolist())
len(temp)
```

```python
y_quant=[]
for i in range(1,101):
    y_quant.append(np.percentile(temp,i))
```

```python
samples=np.random.normal(loc=0,scale=1,size=1000)#mean=0,standard deviation= 1,data_size=1000
```

```python
x_quant=[]
for i in range(1,101):
    x_quant.append(np.percentile(samples,i))
```

```python
sns.scatterplot(x=x_quant,y=y_quant)
```

```python
import statsmodels.api as sm
```

```python
fig=sm.qqplot(df["sepal_length"],line="45",fit=True)# dataset,line(which line to choose),
```
