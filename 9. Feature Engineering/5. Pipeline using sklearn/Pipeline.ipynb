# Pipelines using Sklearn


### Pipelines chain multiple steps so that out of each steps is used as input for next step.

### Pipelines make it easy to apply the same preprocessing to train and test

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("train.csv")
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
```

```python
df.head()
```

```python
df.drop(columns=["PassengerId","Name","Ticket","Cabin"],inplace=True)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(
    df.drop(columns=["Survived"]),
    df["Survived"],
    test_size=0.2,
    random_state=42
)
```

```python
df.isnull().sum()
```

```python
# si_age=SimpleImputer()
# si_embarked
```

```python
from sklearn.compose import ColumnTransformer
```

```python
ct=ColumnTransformer(transformers=[
    ("tnf1",SimpleImputer(strategy="most_frequent"),["Embarked"]),
    ("tnf2",SimpleImputer(),["Age"]),
    ("tnf3",OneHotEncoder(drop="first",sparse_output=False,handle_unknown="ignore",dtype=int),["Sex","Embarked"]),
],remainder="passthrough")
```

```python
x_train_tranform=ct.fit_transform(x_train)
```

```python
df2=pd.DataFrame(x_train_tranform,columns=ct.get_feature_names_out(),index=x_train.index)
```
