```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import Binarizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score 
from sklearn.compose import ColumnTransformer
```

```python
df=pd.read_csv("train.csv",usecols=["Age","Fare","SibSp","Parch","Survived"])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df["Family"]=df["SibSp"]+df["Parch"]
```

```python
df.drop(columns=["SibSp","Parch"],inplace=True)
```

```python
x=df.drop(columns=["Survived"])
y=df["Survived"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    random_state=42,
    test_size=0.2
)
```

```python
clf=DecisionTreeClassifier()
clf.fit(x_train,y_train)
y_pred=clf.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(),x,y,scoring="accuracy",cv=10))
```

```python
trf=ColumnTransformer([
    ("bin",Binarizer(threshold=0,copy=False),["Family"])
],remainder="passthrough")
```

### In Binerizer we have only 2 parameters which threshold (this is the value from which it will compare all other value and assign binary value (0,1)) and copy (this decide weather we want perform on current column or create a copy and perform the operation)

```python
x_train_trf=trf.fit_transform(x_train)
x_test_trf=trf.fit_transform(x_test)
```

```python
pd.DataFrame(x_train_trf,columns=["Family","Age","Fare"])
```

```python
dtc=DecisionTreeClassifier()
dtc.fit(x_train_trf,y_train)
y_pred2=dtc.predict(x_test_trf)
accuracy_score(y_test,y_pred2)
```

```python
x_trf=trf.fit_transform(x)
np.mean(cross_val_score(DecisionTreeClassifier(),x_trf,y,scoring="accuracy",cv=10))
```

### As we can see that there is not much a change in this accuracy using is binerization which is maybe because of dataset but its helpful where we need convert a image to black and white so we set threshold and then perform.
