## Mixed Variable handle where there is mixed data in a column not on same data cell

```python
df["number"].unique()
```

```python
fig=df["number"].value_counts().plot.bar()
fig.set_title("Passengers travelling with")
```

### So here we use the pandas function to numeric for numerical value and replace the non numerical value with NaN and converted in integer

```python
df["numerical_number"]=pd.to_numeric(df["number"],errors="coerce",downcast="integer")
```

### np.where(condition, value_if_true, value_if_false) 
### here this takes  categoraial data in add to other column 

```python
df["categorical_number"]=np.where(df["numerical_number"].isnull(),df["number"],np.nan)
```

## Mixed Variable handle where there is mixed data in a particulat data cell like C82 etc.

```python
df["Cabin"].unique()
```

```python
df["Ticket"].unique()
```

```python
df["num_cabin"]=df["Cabin"].str.extract("(\d+)")#Capturs numerical part
df["cat_cabin"]=df["Cabin"].str[0]# captures the first string 
```

```python
df.head()
```

```python
df["cat_cabin"].value_counts()
```

```python
df["ticket_num"]=df["Ticket"].apply(lambda s: s.split()[-1])
df["ticket_num"]=pd.to_numeric(df["ticket_num"],errors="coerce",downcast="integer")

df["ticket_cat"]=df["Ticket"].apply(lambda s: s.split()[0])
df["ticket_cat"]=np.where(df["ticket_cat"].str.isdigit(),np.nan,df["ticket_cat"])
```
