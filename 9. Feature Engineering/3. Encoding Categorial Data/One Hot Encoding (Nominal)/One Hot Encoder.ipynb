## OneHotEncoding for most frequent categories

```python
count=df["brand"].value_counts()
```

### This wiil take all categoraial values total for each category

```python
df["brand"].nunique()
```

```python
threshold=100
```

```python
repl=count[count<=threshold].index
```

```python
pd.get_dummies(df["brand"].replace(repl,"uncommon"),dtype=int)
```

### Here we set a threshold/limit where the brands with values less than 100 will replace to column name uncommon
