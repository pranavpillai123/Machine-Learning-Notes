```python
date.info()
```

```python
date["date_year"]=date['date'].dt.year
date['date_month']=date['date'].dt.month
date['date_month_name']=date['date'].dt.month_name()
date['date_day']=date['date'].dt.day
date['date_week']=date["date"].dt.dayofweek
date['date_day']=date['date'].dt.day
date['date_date_name']=date['date'].dt.day_name()
```

### This custom fuctions helps to extract some feature like weekend and we can also create our custom functions

```python
date["date_weekend"]=np.where(date["date_week"].isin(["Saturday","Sunday"]),1,0)
date.drop(columns=["product_id","city_id","orders"]).head()
```

```python
date["date_week"]=date["date"].dt.isocalendar().week
```
