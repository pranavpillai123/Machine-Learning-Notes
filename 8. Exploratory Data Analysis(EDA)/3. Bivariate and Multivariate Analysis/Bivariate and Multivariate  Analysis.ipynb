# Scatter plot (Numerical-Numerical)

```python
sns.scatterplot(x=tips["total_bill"],y=tips["tip"],hue=tips["sex"],style=tips["smoker"],size=tips["size"])
```

### As we can see from above Graph without parameter like hue,style and size its a bivariate analysis but with hue,style and size its a multivariate analysis.

# Bar Plot (Numerical-Categorical)

```python
titanic.sample(5)
```

```python
sns.barplot(x=titanic["Pclass"],y=titanic["Age"],hue=titanic["Sex"])
```

### As we can see from above barplot we doing bivariate analysis without hue and using hue we are using multivariate Analysis

# Box Plot (Numerical-Categorical)

```python
sns.boxplot(x=titanic["Sex"],y=titanic["Age"],hue=titanic["Survived"])
```

### As we can see in above box plot we can see its a multivariate analysis as x,y represent age and sex which is bivariate analysis and by adding hue its multivariate analysis

# DistPlot (Numerical-Catergorical)

```python
sns.distplot(titanic[titanic["Survived"]==0]["Age"],hist=False)
sns.distplot(titanic[titanic["Survived"]==1]["Age"],hist=False)
```

### Here we can see that we did bivariate analysis. We can see that probablity of surviving and not surviving in a same graph using kde and pdf

# Heatmap (Categorical-Catergorical)

```python
titanic.head()
```

```python
pd.crosstab(titanic["Pclass"],titanic["Survived"])#This gives info about 2 categorical information
```

```python
sns.heatmap(pd.crosstab(titanic["Pclass"],titanic["Survived"]))
```

### Above graph is bivariate analysis graph where we understand relation of 2 categories using heatmap

```python
titanic.groupby("Pclass")["Survived"].mean()*100 #This shows percentage of survivebility
```

```python
(titanic.groupby("Sex")["Survived"].mean()*100).plot(kind='bar')
```

# Cluster Map (Categorical-Categorical)

```python
pd.crosstab(titanic["Parch"],titanic["Survived"])
```

```python
sns.clustermap(pd.crosstab(titanic["Parch"],titanic["Survived"]))
```

### Above graph is a Clustermap where it does bivariate analysis on dataset

# Line Plot (Numerical-Numerical)

```python
flights.sample(5)
```

```python
fl=flights.groupby("year")["passengers"].sum().reset_index()
```

```python
sns.lineplot(x=fl['year'],y=fl['passengers'])
```

### Above is bivariate analysis in lineplot
