# Numerical Data


## a. Histogram

```python
plt.hist(df["Age"])
```

```python
plt.hist(df["Age"],bins=25)
```

## b. Distplot

```python
sns.distplot(df["Age"])
```

### This tells us the probablity in distplot(older seaborn version)(pdf(probablity density function)) and shows skewness where as hist tells us exact numbers.
### for example in distplot we see probablity of a person being 30 is 1.5 percent. 
### Dist also a KDE(kernal density estimate) used in newer version of seaborn given as below 

```python
sns.histplot(df["Age"],kde=True)
```
